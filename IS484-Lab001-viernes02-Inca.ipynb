```python
# Bloque 1: Importación de librerías
Primero se importan las librerías que se utilizarán. NumPy se usa para generar los datos, Pandas para organizarlos en una tabla y Matplotlib para mostrar los resultados mediante un gráfico.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 

# Bloque2: Generación de transacciones
En este bloque se generan 300 transacciones ficticias o aleatorios por asi decirlo. Para cada transacción se crea una hora del día entre 0 y 23, ademas un monto aleatorio.
La instrucción `np.random.seed(42)` permite que cada vez que se ejecute el programa se obtengan los mismos datos aleatorios, lo cual ayuda a mantener los mismos resultados durante las pruebas.

np.random.seed(42)
n = 300

hora = np.random.randint(0, 24, n)
monto = np.random.exponential(scale=150, size=n) + 10

# Bloque 3: Fraude mas probable de madrugada (0-5h, 23h) y en montos altos
En este bloque se calcula la posibilidad de que una transacción sea fraudulenta.
El fraude existe cuando una mayor posibilidad de la transacción se realiza durante la madrugada o cuando el monto supera los 400 soles.

prob_fraude = 0.03 + 0.35 * ((hora <= 5) | (hora >= 23)) + 0.25 * (monto > 400)
prob_fraude = np.clip(prob_fraude, 0, 0.9)

es_fraude = (np.random.rand(n) < prob_fraude).astype(int)

# Bloque 4: Crear la tabla con los datos
Los datos generados se guardan en una tabla utilizando Pandas. La tabla contiene la hora de la transacción, el monto y una columna que indica si la transacción fue considerada fraude o no.

transacciones = pd.DataFrame({
    "hora": hora,
    "monto": monto,
    "es_fraude": es_fraude
})

# Bloque 5: Colores para identificar las transacciones
luego para diferenciar si es fraude o no asignamos colores. Si es fraude se marcara de color rojo con el codigo de #d03b3b y para saber si es legitimo se marcara con el color verde con el codigo de #0ca30c

colores = {0: "#0ca30c", 1: "#d03b3b"}
etiquetas = {0: "Legitima", 1: "Fraude"}


# Bloque 6: Visualización de las transacciones
En esta parte del codigo nos muestra un grafico ya con todos los datos, es decir los puntos. Cada punto representa una transaccion.
El eje horizontal representa la hora del día y el eje vertical representa el monto de la transacción.
Al momento de correr el codigo completo nos mostrara puntos de color verde que seran las transacciones legitimas y los puntos rojos seran las transacciones fraudes.

fig, ax = plt.subplots(figsize=(10, 6), facecolor="#fcfcfb")
ax.set_facecolor("#fcfcfb")

for valor in [0, 1]:
    subset = transacciones[transacciones["es_fraude"] == valor]

    ax.scatter(
        subset["hora"],
        subset["monto"],
        s=70,
        alpha=0.75,
        c=colores[valor],
        edgecolors="white",
        linewidths=0.6,
        label=etiquetas[valor],
    )

ax.set_title("Transacciones por hora y monto", fontsize=15, color="#0b0b0b", pad=14)
ax.set_xlabel("Hora del dia", fontsize=11, color="#52514e")
ax.set_ylabel("Monto (S/)", fontsize=11, color="#52514e")
ax.set_xticks(range(0, 24, 2))
ax.grid(True, color="#e1e0d9", linewidth=0.8)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(title="Tipo de transaccion", frameon=False, loc="upper right")

plt.tight_layout()
plt.show()

# Bloque 7: Salida
Al ejecutar el codigo completo nos permitira visualizar todo lo programado y explicado anteriormente, es decir nos muestra la presencia de transacciones que son fradudes y las transacciones que son legitimas

print(
    "Total transacciones:", len(transacciones),
    "| Fraudes:", transacciones["es_fraude"].sum()
)
```
